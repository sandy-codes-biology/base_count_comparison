# Base Count Comparison
This notebook takes two FASTA files and reports:
- 🧬 A, T, C and G counts for each sequence
- 📊 A side-by-side bar chart comparing the two sequences

To use your own files, change the four settings lines in the first code cell.

```python
# Settings : change these four lines to use your own files
file1 = "Insulin.fasta"
file2 = "Haemoglobin.fasta"
name1 = "Insulin"
name2 = "Haemoglobin"


def read_fasta(filename):
    sequence = ""
    with open(filename) as f: # Open the file and call it f. It closes by itself when the indented part ends
        for line in f: 
            if not line.startswith(">"):
                sequence += line.strip().upper()
    return sequence

# runs the machine then stores the value under the variable
seq1 = read_fasta(file1)
seq2 = read_fasta(file2)

# Count each base in seq1 
a1 = seq1.count("A")
t1 = seq1.count("T")
c1 = seq1.count("C")
g1 = seq1.count("G")

# Count each base in seq2 
a2 = seq2.count("A")
t2 = seq2.count("T")
c2 = seq2.count("C")
g2 = seq2.count("G")

print(a1, t1, c1, g1)
print(a2, t2, c2, g2)
```

```python
import matplotlib.pyplot as plt

bases = ["A","T","C","G"]
counts1 = [a1, t1, c1, g1]
counts2 = [a2, t2, c2, g2]

positions1 = [0, 1, 2, 3]
positions2 = [0.4, 1.4, 2.4, 3.4]

plt.bar(positions1, counts1, width=0.4, label=name1)
plt.bar(positions2, counts2, width=0.4, label=name2)
# plt.bar draws a bar chart => plt.bar(where, heights) => 'where' could be defined by text or numbers

plt.xticks([0.2, 1.2, 2.2, 3.2], bases) # sets the current x-axis tick positions and their corresponding text labels on a plot
plt.title("Base Counts: " + name1 + " vs " + name2)
plt.xlabel("Base")
plt.ylabel("Count")
plt.legend()
plt.show()
```
